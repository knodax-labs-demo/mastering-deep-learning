```python
import matplotlib.pyplot as plt
import pandas as pd
 
# Create the data
data = {
    'Sunlight (hours)': [1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5, 5.5, 6, 6.5, 7, 7.5, 8, 8.5, 9, 9.5, 10],
    'Water (liters)': [1, 1.25, 1.5, 1.75, 2, 2.25, 2.5, 2.75, 3, 3.25, 3.5, 3.75, 4, 4.25, 4.5, 4.75, 5, 5.25, 5.5],
    'Plant Thrives?': ['Wilted']*6 + ['Thrives']*13
}
 
# Create a DataFrame
df = pd.DataFrame(data)
 
# Plot the data
plt.figure(figsize=(10, 6))
for i in range(len(df)):
    color = 'red' if df['Plant Thrives?'][i] == 'Wilted' else 'blue'
    plt.scatter(df['Sunlight (hours)'][i], df['Water (liters)'][i], color=color, s=80)
 
# Add decision boundary line (example line, can be adjusted)
x_vals = [0, 10]
y_vals = [0.5 * x + 0.5 for x in x_vals]
plt.plot(x_vals, y_vals, color='black', linestyle='--', label='Decision Boundary')
 
# Chart settings
plt.xlabel('Sunlight hours (X1)')
plt.ylabel('Water liters (X2)')
plt.title('Plant Thriving Prediction: Sunlight vs Water')
plt.grid(True)
plt.show()
```
