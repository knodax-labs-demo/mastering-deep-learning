```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import set_matplotlib_formats

# Display plots in SVG format for better quality in Jupyter Notebook
set_matplotlib_formats('svg')

# Define the function
def fx(x):
    """
    Function:
    f(x) = 2x² + 5x + 3
    """
    return 2 * x**2 + 5 * x + 3

# Define the derivative
def deriv(x):
    """
    Derivative:
    f'(x) = 4x + 5
    """
    return 4 * x + 5

# Generate x values
x = np.linspace(-6, 4, 2001)

# Plot the function and derivative
plt.figure(figsize=(8, 5))

plt.plot(x, fx(x), label="f(x) = 2x² + 5x + 3")
plt.plot(x, deriv(x), label="f'(x) = 4x + 5")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Function and Its Derivative")
plt.grid(True)
plt.legend()

plt.show()
```

```python
# Initialize gradient descent parameters

# Random starting point
localmin = np.random.choice(x, 1)

# Learning rate
learning_rate = 0.02

# Number of training iterations
training_epochs = 80

# Store optimization progress
modelparams = np.zeros((training_epochs, 2))

# Run gradient descent
for i in range(training_epochs):

    gradient = deriv(localmin)

    localmin = localmin - learning_rate * gradient

    modelparams[i, 0] = localmin[0]
    modelparams[i, 1] = gradient[0]

print(f"Final estimated local minimum: {localmin[0]:.5f}")
print(f"Function value at minimum: {fx(localmin)[0]:.5f}")
print(f"Gradient at minimum: {deriv(localmin)[0]:.5f}")
```

```python
# Plot the final minimum point

plt.figure(figsize=(8, 5))

plt.plot(x, fx(x), label="f(x)")
plt.plot(x, deriv(x), label="f'(x)")

plt.plot(localmin, fx(localmin), "ro", label="Minimum on f(x)")
plt.plot(localmin, deriv(localmin), "go", label="Gradient near minimum")

plt.xlabel("x")
plt.ylabel("y")
plt.title(f"Estimated Local Minimum: x = {localmin[0]:.5f}")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# Plot the evolution of the local minimum and gradient

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# Plot local minimum evolution
ax[0].plot(modelparams[:, 0], "o-")
ax[0].set_xlabel("Iteration")
ax[0].set_ylabel("Estimated x-value")
ax[0].set_title("Local Minimum Evolution")
ax[0].grid(True)

# Plot gradient evolution
ax[1].plot(modelparams[:, 1], "o-")
ax[1].set_xlabel("Iteration")
ax[1].set_ylabel("Gradient")
ax[1].set_title("Gradient Evolution")
ax[1].grid(True)

plt.tight_layout()
plt.show()
```
