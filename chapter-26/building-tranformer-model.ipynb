# Using nn.MultiheadAttention

```python
import torch
import torch.nn as nn
 
embed_dim = 64
num_heads = 8
 
attention = nn.MultiheadAttention(
    embed_dim=embed_dim,
    num_heads=num_heads,
    batch_first=True
)
 
x = torch.rand(2, 10, 64)
 
output, weights = attention(x, x, x)
 
print("Output Shape:", output.shape)
print("Attention Shape:", weights.shape)
 
```

# Building a Transformer Encoder

```python
import torch
import torch.nn as nn
 
encoder_layer = nn.TransformerEncoderLayer(
    d_model=64,
    nhead=8,
    batch_first=True
)
 
encoder = nn.TransformerEncoder(
    encoder_layer,
    num_layers=2
)
 
x = torch.rand(2, 10, 64)
 
output = encoder(x)
 
print(output.shape)
```

# Building a Transformer Decoder

```python
import torch
import torch.nn as nn
 
decoder_layer = nn.TransformerDecoderLayer(
    d_model=64,
    nhead=8,
    batch_first=True
)
 
decoder = nn.TransformerDecoder(
    decoder_layer,
    num_layers=2
)
 
target = torch.rand(2, 5, 64)
memory = torch.rand(2, 10, 64)
 
output = decoder(target, memory)
 
print(output.shape)
```

# Running Transformer Inference

```python
import torch
import torch.nn as nn
 
model = nn.Transformer(
    d_model=64,
    nhead=8,
    num_encoder_layers=2,
    num_decoder_layers=2,
    batch_first=True
)
 
src = torch.rand(2, 10, 64)
tgt = torch.rand(2, 5, 64)
 
output = model(src, tgt)
 
print(output.shape)
 
```
