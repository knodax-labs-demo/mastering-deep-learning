```python
# =====================================================
# Simple AI Assistant Without External Model
# =====================================================

def generate_response(question):
    question = question.lower()

    if "transfer learning" in question:
        return "Transfer learning reuses a pretrained model for a new but related task."

    elif "cnn" in question:
        return "A CNN is a neural network commonly used for image recognition and computer vision."

    elif "llm" in question:
        return "An LLM is a large language model trained to understand and generate text."

    elif "rag" in question:
        return "RAG combines document retrieval with an LLM to produce more accurate answers."

    else:
        return "I am a simple AI assistant. Ask me about transfer learning, CNNs, LLMs, or RAG."

question = "What is transfer learning?"
answer = generate_response(question)

print("User:", question)
print("Assistant:", answer)
```
