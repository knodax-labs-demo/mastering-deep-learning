# Performing Simple Tensor Operations

````python
import torch
 
x = torch.tensor([1, 2, 3])
y = torch.tensor([4, 5, 6])
 
addition = x + y
multiplication = x * y
 
print("Addition:", addition)
print("Multiplication:", multiplication)
````

# Performing Matrix Operations

````python
import torch
 
matrix1 = torch.tensor([[1, 2], [3, 4]])
matrix2 = torch.tensor([[5, 6], [7, 8]])
 
result = torch.matmul(matrix1, matrix2)
 
print(result)
````

# Running Computation on GPU

````python
import torch
 
if torch.cuda.is_available():
    device = torch.device("cuda")
    print("GPU is available")
else:
    device = torch.device("cpu")
    print("GPU is not available")
 
tensor = torch.tensor([1, 2, 3]).to(device)
 
print(tensor)
````

# What is a Tensor in PyTorch?

In [PyTorch](https://pytorch.org?utm_source=chatgpt.com), a **tensor** is the primary data structure used to store and process data.
A tensor is similar to a NumPy array but provides additional capabilities such as GPU acceleration and automatic differentiation, which are essential for deep learning.

PyTorch tensors are used to represent:

* numbers,
* vectors,
* matrices,
* images,
* and multi-dimensional datasets.

---

# Types of Tensors

A tensor can have different dimensions.

| Tensor Type               | Dimensions | Example             |
| ------------------------- | ---------- | ------------------- |
| Scalar                    | 0D         | `5`                 |
| Vector                    | 1D         | `[1, 2, 3]`         |
| Matrix                    | 2D         | `[[1,2],[3,4]]`     |
| Higher-Dimensional Tensor | 3D+        | Image or video data |

# Performing Tensor Operations

PyTorch tensors support mathematical operations.

```python
x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([4.0, 5.0, 6.0])

addition = x + y
multiplication = x * y

print("Addition:", addition)
print("Multiplication:", multiplication)
```

### Output

```python
Addition: tensor([5., 7., 9.])
Multiplication: tensor([ 4., 10., 18.])
```

---

# Understanding Tensor Shape

The `shape` attribute shows the dimensions of a tensor.

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(x.shape)
```

### Output

```python
torch.Size([2, 3])
```

This means:

* 2 rows
* 3 columns

# Moving Tensors to GPU

PyTorch can move tensors to a GPU for faster computation.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

x = torch.tensor([1, 2, 3]).to(device)

print(x)
print(device)
```

---

# Automatic Gradient Tracking

PyTorch tensors can automatically track gradients during neural network training.

```python
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)

print(x)
```

### Output

```python
tensor([1., 2., 3.], requires_grad=True)
```

Tensors are the foundation of deep learning in PyTorch because they:
