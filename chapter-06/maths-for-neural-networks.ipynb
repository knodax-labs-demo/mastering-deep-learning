# Scalars, Vectors, and Matrices

```python
import torch
vector = torch.tensor([1, 2, 3, 4])
print(vector)
```

```python
matrix = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])
 
print(matrix)
```

# Matrix Multiplication

```python
A = torch.tensor([
    [1, 2],
    [3, 4]
])
 
B = torch.tensor([
    [5, 6],
    [7, 8]
])
 
result = torch.matmul(A, B)
 
print(result)
```

# Tensors in Deep Learning

```python
tensor = torch.tensor([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
 
print(tensor)
print(tensor.shape)
```

# Tensor Shapes and Dimensions

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])
 
print(x.shape)
```

# Broadcasting Operations

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])
 
y = torch.tensor([10, 20, 30])
 
result = x + y
 
print(result)
```

# Hands-On Tensor Operations in PyTorch

```python
x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([4.0, 5.0, 6.0])
 
 
print(x + y)
print(x * y)
print(torch.mean(x))
print(torch.sum(y))
```

# Chain Rule

```python
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2

y.backward()
print(x.grad)
```
